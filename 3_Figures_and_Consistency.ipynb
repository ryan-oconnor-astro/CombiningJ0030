## Consistency of each input

The consistency probability of input $i$ is the probability that it belongs to the good component:

$$
P_i = \left\langle \frac{p\, g_i(M, C)}{p\, g_i(M, C) + (1 - p)\, b_i(M, C \mid \sigma)} \right\rangle,
$$

averaged over the joint posterior of $(M, C, p, \sigma)$. The responsibility itself is untempered; tempering enters through the posterior used for the average. The common systematic displacement is applied after the responsibilities are set, so it does not change $P_i$.

The two ends of the ranking (Kini26 PDT-U highest, Vinciguerra24 ST+PST lowest) are the same for every $\rho$ prescription. The order of the middle six changes with the prescription.

```python
method.print_P_i(res)
```

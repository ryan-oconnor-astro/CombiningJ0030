## Load datasets in the form of .txt files with columns (M, R), compute KDEs, evaluate KDEs on fine grid (Mass and Radius columns of Table 1).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde, norm
import seaborn as sns
import matplotlib as mpl

# ------------------------------------------------------------
# Model definitions
# ------------------------------------------------------------
MODELS = {
    "ST_PST19": dict(file="datafiles/ST_PST19", color="tab:pink",   label="Riley19: ST+PST"),
    "ST_PDT":   dict(file="datafiles/ST_PDT",   color="tab:blue",   label="Vinciguerra24: ST+PDT"),
    "ST_PST":   dict(file="datafiles/ST_PST",   color="tab:orange", label="Vinciguerra24: ST+PST"),
    "PDT_U":    dict(file="datafiles/PDT_U",    color="tab:green",  label="Vinciguerra24: PDT--U"),
    "ST_U":     dict(file="datafiles/ST_U",     color="tab:cyan",   label="Vinciguerra24: ST--U"),
    "2spot":    dict(file="datafiles/2spot",    color="tab:red",    label="Miller19: 2spot"),
    "3spot":    dict(file="datafiles/3spot",    color="tab:purple", label="Miller19: 3spot"),
    "PDT_U26":  dict(file="datafiles/PDT_U26",  color="tab:olive",  label="Kini26: PDT--U"),
}

# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def normalize_2d(pdf, xgrid, ygrid):
    norm_val = np.trapz(np.trapz(pdf, ygrid, axis=1), xgrid)
    return pdf / norm_val

def credible_intervals(grid, pdf):
    dx = grid[1] - grid[0]
    cdf = np.cumsum(pdf) * dx
    cdf /= cdf[-1]
    quantiles = [0.16, 0.50, 0.84]
    values = np.interp(quantiles, cdf, grid)
    return values

def summarize_2d_pdf(M_grid, R_grid, pdf2d):
    """
    From a normalized 2D PDF, compute marginal medians and simple spread proxies.
    """
    P_M = np.trapz(pdf2d, R_grid, axis=1)
    P_M /= np.trapz(P_M, M_grid)

    P_R = np.trapz(pdf2d, M_grid, axis=0)
    P_R /= np.trapz(P_R, R_grid)

    M_q16, M_q50, M_q84 = credible_intervals(M_grid, P_M)
    R_q16, R_q50, R_q84 = credible_intervals(R_grid, P_R)

    # Keep the same spread-style logic as your original code
    S_M = max(M_q84 - M_q50, M_q50 - M_q16)
    S_R = max(R_q84 - R_q50, R_q50 - R_q16)

    return {
        "M_med": M_q50,
        "M_16": M_q16,
        "M_84": M_q84,
        "R_med": R_q50,
        "R_16": R_q16,
        "R_84": R_q84,
        "S_M": S_M,
        "S_R": S_R,
        "P_M": P_M,
        "P_R": P_R,
    }

# ------------------------------------------------------------
# Load data and compute individual KDEs
# ------------------------------------------------------------
samples_MR = {}
M_med, M_lo, M_hi, M_16, M_84, S_Mi = {}, {}, {}, {}, {}, {}
R_med, R_lo, R_hi, R_16, R_84, S_Ri = {}, {}, {}, {}, {}, {}
KDEs_2D = {}

for key, meta in MODELS.items():
    filename = meta["file"]

    data = np.loadtxt(filename)
    M, R = data.T[0], data.T[1]
    samples_MR[key] = np.vstack([M, R])

    # Mass statistics
    M_med[key] = np.median(M)
    M_lo[key]  = np.min(M)
    M_hi[key]  = np.max(M)
    M_16[key]  = np.percentile(M, 16)
    M_84[key]  = np.percentile(M, 84)
    S_Mi[key]  = max(M_hi[key] - M_med[key], M_med[key] - M_lo[key])

    # Radius statistics
    R_med[key] = np.median(R)
    R_lo[key]  = np.min(R)
    R_hi[key]  = np.max(R)
    R_16[key]  = np.percentile(R, 16)
    R_84[key]  = np.percentile(R, 84)
    S_Ri[key]  = max(R_hi[key] - R_med[key], R_med[key] - R_lo[key])

    print(
        rf"{meta['label']}: "
        rf"M = {M_med[key]:.3f} (+{M_84[key] - M_med[key]:.3f}/-{M_med[key] - M_16[key]:.3f}), "
        rf"R = {R_med[key]:.3f} (+{R_84[key] - R_med[key]:.3f}/-{R_med[key] - R_16[key]:.3f})"
    )

    KDEs_2D[key] = gaussian_kde(samples_MR[key], bw_method="silverman")

# ------------------------------------------------------------
# Grids
# ------------------------------------------------------------
M_grid = np.linspace(0.95, 2.2, 100)
R_grid = np.linspace(9.0, 17.0, 100)

MM, RR = np.meshgrid(M_grid, R_grid, indexing="ij")
pos = np.vstack([MM.ravel(), RR.ravel()])

# ------------------------------------------------------------
# Evaluate and normalize individual KDEs on grid
# ------------------------------------------------------------
g_dict = {}
for key in MODELS:
    g = KDEs_2D[key](pos).reshape(MM.shape)
    g = normalize_2d(g, M_grid, R_grid)
    g_dict[key] = g
```

```python
# ------------------------------------------------------------
# USER-DEFINED CLASS ASSIGNMENTS
# Give each measurement a class label: 1, 2, 3, ...
# Measurements with the same class label will be summed together.
# ------------------------------------------------------------
CLASS_MAP = {
    "ST_PST19": 1,
    "ST_PDT":   1,
    "ST_PST":   1,
    "PDT_U":    1,
    "ST_U":     1,
    "2spot":    2,
    "3spot":    2,
    "PDT_U26":  1,
}

# ------------------------------------------------------------
# BUILD CLASS-LEVEL "GOOD" DISTRIBUTIONS
# Sum all measurements in the same class, then renormalize
# ------------------------------------------------------------
class_ids = sorted(set(CLASS_MAP.values()))
class_members = {cid: [k for k in MODELS if CLASS_MAP[k] == cid] for cid in class_ids}

g_class_dict = {}
class_stats = {}

print("\nClass assignments:")
for cid in class_ids:
    print(f"Class {cid}: {class_members[cid]}")

for cid in class_ids:
    g_class = np.zeros_like(MM)
    for key in class_members[cid]:
        g_class += g_dict[key]

    g_class = normalize_2d(g_class, M_grid, R_grid)
    g_class_dict[cid] = g_class

    stats = summarize_2d_pdf(M_grid, R_grid, g_class)
    class_stats[cid] = stats

    print(
        rf"Class {cid}: "
        rf"M = {stats['M_med']:.3f} (+{stats['M_84'] - stats['M_med']:.3f}/-{stats['M_med'] - stats['M_16']:.3f}), "
        rf"R = {stats['R_med']:.3f} (+{stats['R_84'] - stats['R_med']:.3f}/-{stats['R_med'] - stats['R_16']:.3f})"
    )

# Stack over classes, not measurements
g_stack = np.array([g_class_dict[cid] for cid in class_ids])   # shape: (Nclass, NM, NR)

# ------------------------------------------------------------
# Priors for systematic-error parameters
# Use class-level spreads now
# ------------------------------------------------------------
avg_S_Mi = np.mean([class_stats[cid]["S_M"] for cid in class_ids])
sigma_M_grid = np.logspace(-10 * avg_S_Mi, -0.1 * avg_S_Mi, 10)
prior_sigma_M = 1.0 / sigma_M_grid
prior_sigma_M /= np.trapz(prior_sigma_M, sigma_M_grid)

avg_S_Ri = np.mean([class_stats[cid]["S_R"] for cid in class_ids])
sigma_R_grid = np.logspace(-10 * avg_S_Ri, -0.1 * avg_S_Ri, 10)
prior_sigma_R = 1.0 / sigma_R_grid
prior_sigma_R /= np.trapz(prior_sigma_R, sigma_R_grid)

alpha_grid = np.linspace(1.0, 10.0, 80)
prior_alpha = np.exp(-alpha_grid)
prior_alpha /= np.trapz(prior_alpha, alpha_grid)

p_grid = np.linspace(0.0, 1.0, 40)  # flat prior on p

# ------------------------------------------------------------
# Precompute class-level bad distributions:
# b_c(M,R | sigma_M, sigma_R)
# ------------------------------------------------------------
b_class_dict = {}

for cid in class_ids:
    mu_M = class_stats[cid]["M_med"]
    mu_R = class_stats[cid]["R_med"]
    S_M_class = class_stats[cid]["S_M"]
    S_R_class = class_stats[cid]["S_R"]

    b_sigma = []

    for sigma_M in sigma_M_grid:
        row_sigma_R = []

        for sigma_R in sigma_R_grid:
            b_alpha = []

            for alpha in alpha_grid:
                var_M = (alpha * S_M_class)**2 + sigma_M**2
                var_R = (alpha * S_R_class)**2 + sigma_R**2

                b = (
                    norm.pdf(MM, loc=mu_M, scale=np.sqrt(var_M)) *
                    norm.pdf(RR, loc=mu_R, scale=np.sqrt(var_R))
                )

                b = normalize_2d(b, M_grid, R_grid)
                b_alpha.append(b)

            b_alpha = np.array(b_alpha)

            # Marginalize over alpha
            b_marg = np.trapz(
                b_alpha * prior_alpha[:, None, None],
                alpha_grid,
                axis=0
            )

            b_marg = normalize_2d(b_marg, M_grid, R_grid)
            row_sigma_R.append(b_marg)

        b_sigma.append(row_sigma_R)

    b_class_dict[cid] = np.array(b_sigma)

# ------------------------------------------------------------
# Main class-level combination
# ------------------------------------------------------------
NsM = len(sigma_M_grid)
NsR = len(sigma_R_grid)

evidence_sigma_MR = np.zeros((NsM, NsR))
P_MR_given_sigma = np.zeros((NsM, NsR, len(M_grid), len(R_grid)))

for iM, sigma_M in enumerate(sigma_M_grid):
    for iR, sigma_R in enumerate(sigma_R_grid):

        b_stack = np.array([b_class_dict[cid][iM, iR] for cid in class_ids])

        g_exp = g_stack[:, None, :, :]        # (Nclass, Np, NM, NR)
        b_exp = b_stack[:, None, :, :]        # (Nclass, Np, NM, NR)
        p_exp = p_grid[None, :, None, None]   # (1, Np, 1, 1)

        mixture = p_exp * g_exp + (1.0 - p_exp) * b_exp
        L_p = np.prod(mixture, axis=0)        # (Np, NM, NR)

        likelihood = np.trapz(L_p, p_grid, axis=0)

        Z = np.trapz(np.trapz(likelihood, R_grid, axis=1), M_grid)

        evidence_sigma_MR[iM, iR] = Z
        P_MR_given_sigma[iM, iR] = likelihood / Z

# ------------------------------------------------------------
# Posterior over sigma_M, sigma_R
# ------------------------------------------------------------
posterior_sigma_MR = (
    evidence_sigma_MR
    * prior_sigma_M[:, None]
    * prior_sigma_R[None, :]
)

posterior_sigma_MR /= np.trapz(
    np.trapz(posterior_sigma_MR, sigma_R_grid, axis=1),
    sigma_M_grid
)

# ------------------------------------------------------------
# Final posterior P(M,R)
# ------------------------------------------------------------
posterior = np.trapz(
    np.trapz(
        P_MR_given_sigma * posterior_sigma_MR[:, :, None, None],
        sigma_R_grid,
        axis=1
    ),
    sigma_M_grid,
    axis=0
)

posterior = normalize_2d(posterior, M_grid, R_grid)

print("\n2D M–R combination complete.")

# ------------------------------------------------------------
# Marginal posteriors for M and R
# ------------------------------------------------------------
P_M = np.trapz(posterior, R_grid, axis=1)
P_M /= np.trapz(P_M, M_grid)

P_R = np.trapz(posterior, M_grid, axis=0)
P_R /= np.trapz(P_R, R_grid)

M_q16, M_q50, M_q84 = credible_intervals(M_grid, P_M)
R_q16, R_q50, R_q84 = credible_intervals(R_grid, P_R)

print("\nFinal marginalized constraints:")
print(f"M = {M_q50:.4f}+{M_q84 - M_q50:.4f}/-{M_q50 - M_q16:.4f} SM")
print(f"R = {R_q50:.4f}+{R_q84 - R_q50:.4f}/-{R_q50 - R_q16:.4f} km")

# ------------------------------------------------------------
# Draw samples from posterior and save
# ------------------------------------------------------------
n_samples = 10000
posterior_flat = posterior.ravel()
posterior_flat /= np.sum(posterior_flat)

indices = np.random.choice(len(posterior_flat), size=n_samples, p=posterior_flat)
M_samples = MM.ravel()[indices]
R_samples = RR.ravel()[indices]

np.savetxt(
    "combined_posterior_classmethod.txt",
    np.vstack([M_samples, R_samples]).T,
    header="M R"
)
```

```python
# Flatten posterior
flat = posterior.ravel()
sorted_idx = np.argsort(flat)[::-1]
sorted_pdf = flat[sorted_idx]

# Cumulative probability
cumulative = np.cumsum(sorted_pdf)
cumulative /= cumulative[-1]

# Find thresholds for 68% and 95%
level_68 = sorted_pdf[np.searchsorted(cumulative, 0.68)]
level_95 = sorted_pdf[np.searchsorted(cumulative, 0.95)]

plt.style.use('seaborn-v0_8-white')
plt.rc('text', usetex=True)
plt.rc('font', family='serif')

# HDR level
def compute_hdr_levels(pdf, probs=(0.68, 0.95)):

    flat = pdf.ravel()
    idx = np.argsort(flat)[::-1]
    sorted_pdf = flat[idx]

    cumsum = np.cumsum(sorted_pdf)
    cumsum /= cumsum[-1]

    levels = []
    for p in probs:
        levels.append(sorted_pdf[np.searchsorted(cumsum, p)])

    return levels

# ------------------------------------------------------------
# Plot: class posteriors vs combined posterior (same style)
# ------------------------------------------------------------

# Compute posterior HDR levels
posterior_lvl68, posterior_lvl95 = compute_hdr_levels(posterior, (0.68, 0.95))

# Build class comparison list
compare_classes = [
    (cid, f"Class {cid}: {', '.join(class_members[cid])}")
    for cid in class_ids
]

# Assign colors to classes (distinct but simple)
class_colors = {
    cid: mpl.cm.tab10(i % 10)
    for i, cid in enumerate(class_ids)
}

# Grid layout
n_classes = len(compare_classes)
ncols = 3
nrows = int(np.ceil((n_classes + 1) / ncols))  # +1 for "all classes" panel

fig, axes = plt.subplots(nrows, ncols, figsize=(24, 6 * nrows), constrained_layout=True)
axes = axes.flatten()

def rgba(color, alpha):
    return mpl.colors.to_rgba(color, alpha)

# ------------------------------------------------------------
# Panels: each class vs posterior
# ------------------------------------------------------------
for ax, (cid, label) in zip(axes[:n_classes], compare_classes):

    density = g_class_dict[cid]
    lvl68, lvl95 = compute_hdr_levels(density, (0.68, 0.95))

    color = class_colors[cid]

    # Posterior fill
    ax.contourf(
        R_grid, M_grid, posterior,
        levels=[posterior_lvl95, posterior_lvl68, posterior.max()],
        colors=[rgba("black", 0.10), rgba("black", 0.25)],
        antialiased=True
    )

    # Class fill
    ax.contourf(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68, density.max()],
        colors=[rgba(color, 0.20), rgba(color, 0.45)],
        antialiased=True
    )

    # Posterior outline
    ax.contour(
        R_grid, M_grid, posterior,
        levels=[posterior_lvl95, posterior_lvl68],
        colors=rgba("black", 0.7),
        linewidths=[1.8, 2.5],
        linestyles=['--', '-']
    )

    # Class outline
    ax.contour(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68],
        colors=rgba(color, 0.7),
        linewidths=[1.3, 1.8],
        linestyles=['--', '-'],
    )

    ax.set_title(label, fontsize=20)
    ax.set_xlabel("Radius (km)", fontsize=18)
    ax.set_ylabel("Mass ($M_\\odot$)", fontsize=18)
    ax.minorticks_on()
    ax.tick_params(labelsize=12)

    ax.legend(
        handles=[
            mpl.lines.Line2D([], [], color=color, linewidth=2, label=f"Class {cid}"),
            mpl.lines.Line2D([], [], color='black', linewidth=2, label='Posterior')
        ],
        fontsize=16,
        loc='upper left',
        frameon=True,
        fancybox=True,
        framealpha=0.8
    )

# ------------------------------------------------------------
# Final panel: all classes + posterior
# ------------------------------------------------------------
ax_all = axes[n_classes]

for cid, label in compare_classes:

    density = g_class_dict[cid]
    lvl68, lvl95 = compute_hdr_levels(density, (0.68, 0.95))
    color = class_colors[cid]

    # Class fill
    ax_all.contourf(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68, density.max()],
        colors=[rgba(color, 0.10), rgba(color, 0.2)],
        antialiased=True
    )
    
    # Class outline
    ax_all.contour(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68],
        colors=rgba(color, 0.7),
        linewidths=[1.3, 1.8],
        linestyles=['--', '-'],
    )

# Posterior fill
ax_all.contourf(
    R_grid, M_grid, posterior,
    levels=[posterior_lvl95, posterior_lvl68, posterior.max()],
    colors=[rgba("black", 0.10), rgba("black", 0.25)],
    antialiased=True
)

# Posterior outline
ax_all.contour(
    R_grid, M_grid, posterior,
    levels=[posterior_lvl95, posterior_lvl68],
    colors='black',
    linewidths=[1.8, 2.5],
    linestyles=['--', '-']
)

ax_all.set_title("All Classes", fontsize=20)
ax_all.set_xlabel("Radius (km)", fontsize=18)
ax_all.set_ylabel("Mass ($M_\\odot$)", fontsize=18)
ax_all.minorticks_on()
ax_all.tick_params(labelsize=12)

# Hide unused panels
for ax in axes[n_classes+1:]:
    ax.axis("off")

fig.suptitle(
    "Class M--R Posteriors Compared to Posterior Combination",
    fontsize=28
)

plt.show()
```

```python
# Flatten posterior
flat = posterior.ravel()
sorted_idx = np.argsort(flat)[::-1]
sorted_pdf = flat[sorted_idx]

# Cumulative probability
cumulative = np.cumsum(sorted_pdf)
cumulative /= cumulative[-1]

# Find thresholds for 68% and 95%
level_68 = sorted_pdf[np.searchsorted(cumulative, 0.68)]
level_95 = sorted_pdf[np.searchsorted(cumulative, 0.95)]

plt.style.use('seaborn-v0_8-white')
plt.rc('text', usetex=True)
plt.rc('font', family='serif')

# HDR level
def compute_hdr_levels(pdf, probs=(0.68, 0.95)):

    flat = pdf.ravel()
    idx = np.argsort(flat)[::-1]
    sorted_pdf = flat[idx]

    cumsum = np.cumsum(sorted_pdf)
    cumsum /= cumsum[-1]

    levels = []
    for p in probs:
        levels.append(sorted_pdf[np.searchsorted(cumsum, p)])

    return levels


# Posterior levels
posterior_lvl68, posterior_lvl95 = compute_hdr_levels(posterior, (0.68, 0.95))

# Models to compare
compare_series = [
    (key, MODELS[key]["label"])
    for key in MODELS
]

#3x3
fig, axes = plt.subplots(3, 3, figsize=(24, 18), constrained_layout=True)
axes = axes.flatten()

def rgba(color, alpha):
    return mpl.colors.to_rgba(color, alpha)


# Panels 1–7: individual vs posterior
for ax, (key, label) in zip(axes[:len(compare_series)], compare_series):

    density = g_dict[key]
    lvl68, lvl95 = compute_hdr_levels(density, (0.68, 0.95))

    # Posterior fill
    ax.contourf(
        R_grid, M_grid, posterior,
        levels=[posterior_lvl95, posterior_lvl68, posterior.max()],
        colors=[rgba("black", 0.10), rgba("black", 0.25)],
        antialiased=True
    )

    # Individual fill
    ax.contourf(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68, density.max()],
        colors=[rgba(MODELS[key]["color"], 0.20), rgba(MODELS[key]["color"], 0.45)],
        antialiased=True
    )

    # Posterior outline
    ax.contour(
        R_grid, M_grid, posterior,
        levels=[posterior_lvl95, posterior_lvl68],
        colors=rgba("black", 0.7),
        linewidths=[1.8, 2.5],
        linestyles=['--', '-']
    )

    # Individual outline
    ax.contour(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68],
        colors=rgba(MODELS[key]["color"], 0.7),
        linewidths=[1.3, 1.8],
        linestyles=['--', '-'],
    )

    ax.set_title(label, fontsize=26)
    ax.set_xlabel("Radius (km)", fontsize=22)
    ax.set_ylabel("Mass ($M_\\odot$)", fontsize=22)
    ax.minorticks_on()
    ax.tick_params(labelsize=14)

    ax.legend(
        handles=[
            mpl.lines.Line2D([], [], color='tab:blue', linewidth=2, label=label),
            mpl.lines.Line2D([], [], color='black', linewidth=2, label='Posterior')
        ],
        fontsize=22,
        loc='upper left',
        frameon=True,
        fancybox=True,
        framealpha=0.8
    )

# Panel 8: all models + posterior
ax_all = axes[-1]

for key, label in compare_series:

    density = g_dict[key]
    lvl68, lvl95 = compute_hdr_levels(density, (0.68, 0.95))

    # Individual fill
    ax_all.contourf(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68, density.max()],
        colors=[rgba(MODELS[key]["color"], 0.10), rgba(MODELS[key]["color"], 0.2)],
        antialiased=True
    )
    
    # Individual outline
    ax_all.contour(
        R_grid, M_grid, density,
        levels=[lvl95, lvl68],
        colors=rgba(MODELS[key]["color"], 0.7),
        linewidths=[1.3, 1.8],
        linestyles=['--', '-'],
    )

# Posterior fill
ax_all.contourf(
    R_grid, M_grid, posterior,
    levels=[posterior_lvl95, posterior_lvl68, posterior.max()],
    colors=[rgba("black", 0.10), rgba("black", 0.25)],
    antialiased=True
)

# Posterior outline
ax_all.contour(
    R_grid, M_grid, posterior,
    levels=[posterior_lvl95, posterior_lvl68],
    colors='black',
    linewidths=[1.8, 2.5],
    linestyles=['--', '-']
)

ax_all.set_title("All Models", fontsize=26)
ax_all.set_xlabel("Radius (km)", fontsize=22)
ax_all.set_ylabel("Mass ($M_\\odot$)", fontsize=22)
ax_all.minorticks_on()
ax_all.tick_params(labelsize=14)

fig.suptitle(
    "Model M--R Posteriors Compared to Posterior Combination",
    fontsize=36
)

plt.show()
posterior_2d = posterior  # for naming consistency


# Marginal distributions

# P(R) = ∫ P(M,R) dM
P_R = np.trapz(posterior_2d, M_grid, axis=0)
P_R /= np.trapz(P_R, R_grid)

# P(M) = ∫ P(M,R) dR
P_M = np.trapz(posterior_2d, R_grid, axis=1)
P_M /= np.trapz(P_M, M_grid)

# Quantiles (16%, 50%, 84%) — properly normalized
def compute_quantiles(grid, pdf):
    cdf = np.cumsum(pdf) * (grid[1] - grid[0])
    cdf /= cdf[-1]
    return np.interp([0.16, 0.5, 0.84], cdf, grid)

R_q16, R_q50, R_q84 = compute_quantiles(R_grid, P_R)
M_q16, M_q50, M_q84 = compute_quantiles(M_grid, P_M)
```

## 'Credibility' score (Table 2, Section 3.3)

```python
# Score each individual measurement against the 2‑D posterior
# (approximate by marginalizing the bad component over the σ_M,σ_R posterior)
epsilon = 1e-300  # Tiny value to avoid divide-by-zero

# precompute marginal bad pdf for each dataset by integrating over sigma_M and sigma_R
b_marginal = {}
for key in MODELS:
    # b_dict[key] has shape (NsM, NsR, NM, NR)
    # weight with posterior_sigma_MR and integrate over both sigma axes
    # result has shape (NM, NR)
    weighted = b_dict[key] * posterior_sigma_MR[:, :, None, None]
    b_m = np.trapz(
            np.trapz(weighted, sigma_R_grid, axis=1),
            sigma_M_grid,
            axis=0
        )
    # normalize on M,R grid just to be safe
    norm = np.trapz(np.trapz(b_m, R_grid, axis=1), M_grid)
    if norm > 0:
        b_m /= norm
    b_marginal[key] = b_m

# iterate over the mass/radius grid
P_vi_1 = {}
for name in MODELS:
    prob_grid = np.zeros_like(posterior)  # (NM,NR)
    for i in range(len(M_grid)):
        for j in range(len(R_grid)):
            f_good = g_dict[name][i, j]
            f_bad_value = b_marginal[name][i, j]
            p_term = (p_grid * f_good) / (
                p_grid * f_good + (1 - p_grid) * f_bad_value + epsilon
            )
            prob_grid[i, j] = 1000*np.trapz(p_term, p_grid) * posterior[i, j]
    # integrate over M,R
    P_vi_1[name] = np.trapz(np.trapz(prob_grid, R_grid, axis=1), M_grid)
    print(f"Probability of {name} being good: {P_vi_1[name]:.4f}")
```
